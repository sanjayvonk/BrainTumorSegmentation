```python
import pyvista as pv
import numpy as np

# Use the 'client' backend for interactive plotting in Jupyter
pv.set_jupyter_backend('client')

# --------------------------
# Dummy Data Setup (Replace with your actual data)
# --------------------------
IMG_SIZE = 64
VOLUME_SLICES = 64

# Create coordinate arrays spanning [-1, 1] in each dimension.
xs = np.linspace(-1, 1, IMG_SIZE)
ys = np.linspace(-1, 1, IMG_SIZE)
zs = np.linspace(-1, 1, VOLUME_SLICES)
X, Y, Z = np.meshgrid(xs, ys, zs, indexing='ij')

# Dummy volume data: a simple radial field (replace this with your volume data, e.g. origImage)
origImage = np.sqrt(X**2 + Y**2 + Z**2)

# --------------------------
# Create a Structured Grid from the 3D Data
# --------------------------
def create_structured_grid(data):
    nx, ny, nz = data.shape
    xs = np.linspace(-1, 1, nx)
    ys = np.linspace(-1, 1, ny)
    zs = np.linspace(-1, 1, nz)
    X, Y, Z = np.meshgrid(xs, ys, zs, indexing='ij')
    # Stack the coordinates into an (n_points, 3) array
    points = np.column_stack((X.ravel(), Y.ravel(), Z.ravel()))
    
    grid = pv.StructuredGrid()
    grid.points = points
    grid.dimensions = data.shape  # (nx, ny, nz)
    grid["values"] = data.flatten(order="F")  # For VTK compatibility, use Fortran order
    return grid

grid = create_structured_grid(origImage)

# --------------------------
# Create an Interactive Plotter
# --------------------------
plotter = pv.Plotter(notebook=True, window_size=(800, 600))
plotter.add_volume(grid,
                   scalars="values",
                   opacity="sigmoid",
                   cmap="gray",
                   shade=True)
# Set an initial camera view (isometric)
plotter.camera_position = 'iso'

# --------------------------
# (Optional) Auto-Rotation Callback
# --------------------------
def update_camera():
    # Rotate the camera's azimuth by 1 degree on each callback.
    plotter.camera.Azimuth(1)
    plotter.render()

# Uncomment the following line to enable automatic rotation.
plotter.add_callback(update_camera, interval=50)  # Update every 50 milliseconds

# --------------------------
# Display the Interactive Plot in the Notebook
# --------------------------
plotter.show()
```
